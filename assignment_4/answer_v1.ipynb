## 5. Baseline Text Generation

Generating text continuations using the pretrained DistilGPT-2 model without any fine-tuning. This establishes our baseline performance for comparison with later fine-tuned versions.

```python
def generate_text_samples(model, tokenizer, prompt, num_samples=3, **generation_params):
    """Generate multiple text samples using the given prompt."""
    generated_texts = []
    
    print(f"Generating {num_samples} text samples...")
    print(f"Prompt: {prompt}\\n")
    
    for i in range(num_samples):
        print(f"--- Sample {i+1} ---")
        
        # Tokenize the prompt
        inputs = tokenizer.encode(prompt, return_tensors='pt').to(device)
        
        # Generate text
        start_time = time.time()
        with torch.no_grad():
            outputs = model.generate(inputs, **generation_params)
        
        generation_time = time.time() - start_time
        
        # Decode the generated text
        generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
        
        # Extract only the generated part (remove the prompt)
        generated_continuation = generated_text[len(prompt):].strip()
        
        generated_texts.append({
            'sample_id': i+1,
            'full_text': generated_text,
            'continuation': generated_continuation,
            'generation_time': generation_time,
            'total_tokens': len(outputs[0]),
            'new_tokens': len(outputs[0]) - len(inputs[0])
        })
        
        print(f"Generated continuation: {generated_continuation}")
        print(f"Generation time: {generation_time:.2f}s")
        print(f"Tokens generated: {len(outputs[0]) - len(inputs[0])}")
        print()
    
    return generated_texts

# Generate baseline samples
baseline_samples = generate_text_samples(
    model, tokenizer, GEN_PROMPT, 
    num_samples=3, 
    **generation_params
)
```

## 8. Quantitative Evaluation - Perplexity

Computing perplexity as a quantitative measure of how well the model predicts the text. Lower perplexity indicates better fit to the domain.

```python
def calculate_perplexity(model, tokenizer, texts, device):
    """Calculate perplexity for a list of texts."""
    model.eval()
    total_loss = 0.0
    total_tokens = 0
    
    with torch.no_grad():
        for text in texts:
            inputs = tokenizer.encode(text, return_tensors='pt').to(device)
            
            # skip very short texts
            if inputs.shape[1] < 2:
                continue
                
            outputs = model(inputs, labels=inputs)
            loss = outputs.loss
            
            total_loss += loss.item() * (inputs.shape[1] - 1) 
            total_tokens += inputs.shape[1] - 1
    
    #  average loss and perplexity
    avg_loss = total_loss / total_tokens if total_tokens > 0 else float('inf')
    perplexity = torch.exp(torch.tensor(avg_loss)).item()
    
    return perplexity, avg_loss

print("=== PERPLEXITY EVALUATION ===\\n")

print("Computing perplexity for reference scientific abstracts...")
ref_perplexity, ref_loss = calculate_perplexity(model, tokenizer, scientific_abstracts, device)
print(f"Reference abstracts - Perplexity: {ref_perplexity:.2f}, Loss: {ref_loss:.4f}")

# perplexity for generated samples
generated_texts = [sample['continuation'] for sample in baseline_samples]
print("\\nComputing perplexity for generated samples...")
gen_perplexity, gen_loss = calculate_perplexity(model, tokenizer, generated_texts, device)
print(f"Generated samples - Perplexity: {gen_perplexity:.2f}, Loss: {gen_loss:.4f}")

general_texts = [
    "The weather is nice today and I decided to go for a walk in the park.",
    "Yesterday I went to the store to buy some groceries including milk and bread.",
    "My favorite movie is about a young wizard who goes to a magical school."
]
print("\\nComputing perplexity for general text (for comparison)...")
general_perplexity, general_loss = calculate_perplexity(model, tokenizer, general_texts, device)
print(f"General text - Perplexity: {general_perplexity:.2f}, Loss: {general_loss:.4f}")

print("INTERPRETATION:")
print(f"- Reference abstracts have perplexity of {ref_perplexity:.2f}")
print(f"- Generated samples have perplexity of {gen_perplexity:.2f}")
print(f"- General text has perplexity of {general_perplexity:.2f}")
print()
if ref_perplexity < gen_perplexity:
    print("Reference abstracts have lower perplexity than generated text")
    print("  This indicates the model fits better to authentic scientific text")
else:
    print("Generated text has lower perplexity than reference abstracts")
    print("  This suggests the model generates text closer to its training distribution")
```

# C-Level Task Implementation

## 1. Fine-tuning Preparation

Setting up the environment for fine-tuning the DistilGPT-2 model on our scientific abstracts dataset. We'll prepare the data in the format required for causal language modeling.

```python
# Import fine-tuning libraries
from transformers import TrainingArguments, Trainer, DataCollatorForLanguageModeling
from datasets import Dataset

print("Fine-tuning environment ready!")

# Prepare dataset for training
def prepare_dataset_for_training(texts, tokenizer, max_length=256):
    """Tokenize texts for causal language modeling."""
    tokenized_texts = []
    for text in texts:
        tokens = tokenizer.encode(text, max_length=max_length, truncation=True)
        tokenized_texts.append(tokens)
    
    return Dataset.from_dict({'input_ids': tokenized_texts})

# Create training dataset
training_dataset = prepare_dataset_for_training(scientific_abstracts, tokenizer)

print(f"Training dataset: {len(training_dataset)} samples")
print(f"Example length: {len(training_dataset[0]['input_ids'])} tokens")
```

## 2. Full Fine-tuning Implementation

Implementing full fine-tuning of DistilGPT-2 on our scientific abstracts. We'll use a lightweight training setup suitable for the small dataset size.

```python
# Create a copy of the model for fine-tuning
import copy

print("=== FULL FINE-TUNING SETUP ===")

# Create a fresh copy of the model for fine-tuning
finetuned_model = copy.deepcopy(model)
finetuned_model.train()

# Count trainable parameters
total_params = sum(p.numel() for p in finetuned_model.parameters())
trainable_params = sum(p.numel() for p in finetuned_model.parameters() if p.requires_grad)

print(f"Total parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
print(f"Trainable percentage: {100 * trainable_params / total_params:.1f}%")

# Fine-tuning configuration
training_config = {
    'learning_rate': 5e-5,
    'num_epochs': 3,
    'batch_size': 2,
    'warmup_steps': 10,
    'max_grad_norm': 1.0,
}

print(f"\\nTraining Configuration:")
for key, value in training_config.items():
    print(f"  {key}: {value}")

# Setup training components
data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False,  # Causal language modeling
)

training_args = TrainingArguments(
    output_dir='./finetuned_model',
    overwrite_output_dir=True,
    num_train_epochs=training_config['num_epochs'],
    per_device_train_batch_size=training_config['batch_size'],
    learning_rate=training_config['learning_rate'],
    warmup_steps=training_config['warmup_steps'],
    max_grad_norm=training_config['max_grad_norm'],
    no_cuda=True,
    report_to=None,  # Disable logging
    save_total_limit=1,
)
```

```python
# Create trainer and run fine-tuning
trainer = Trainer(
    model=finetuned_model,
    args=training_args,
    train_dataset=training_dataset,
    data_collator=data_collator,
    tokenizer=tokenizer,
)

print("=== STARTING FINE-TUNING ===")
print(f"Training on {len(training_dataset)} samples for {training_config['num_epochs']} epochs")

# Track training time
import time
training_start_time = time.time()

# Start training
train_result = trainer.train()
training_end_time = time.time()
training_time = training_end_time - training_start_time

print(f"\\n=== TRAINING COMPLETED ===")
print(f"Training time: {training_time:.2f} seconds")
print(f"Final training loss: {train_result.training_loss:.4f}")
print(f"Training steps: {train_result.global_step}")

# Save the fine-tuned model
trainer.save_model()
print("Fine-tuned model saved successfully!")
```

## 6. Custom Modification: Data Augmentation Approach

Implementing a custom modification using **data augmentation** by expanding our training dataset with synthetic variations and training for more epochs with a lower learning rate.

```python
# Custom Modification: Data Augmentation + Modified Training
print("=== CUSTOM MODIFICATION EXPERIMENT ===\\n")

# Create augmented dataset by extracting sentences and creating variations
def create_augmented_dataset(abstracts, augmentation_factor=2):
    """Create an augmented dataset by sentence extraction and recombination."""
    augmented_texts = list(abstracts)  # Start with original abstracts
    
    # Extract sentences from all abstracts
    all_sentences = []
    for abstract in abstracts:
        sentences = [s.strip() for s in abstract.split('.') if s.strip()]
        all_sentences.extend(sentences)
    
    # Create synthetic abstracts by combining sentences
    import random
    random.seed(42)  # For reproducibility
    
    for _ in range(len(abstracts) * augmentation_factor):
        # Select 3-4 sentences randomly
        num_sentences = random.randint(3, 4)
        selected_sentences = random.sample(all_sentences, min(num_sentences, len(all_sentences)))
        synthetic_abstract = '. '.join(selected_sentences) + '.'
        augmented_texts.append(synthetic_abstract)
    
    return augmented_texts

# Create augmented dataset
print("Creating augmented dataset...")
augmented_abstracts = create_augmented_dataset(scientific_abstracts, augmentation_factor=1)
print(f"Original dataset: {len(scientific_abstracts)} samples")
print(f"Augmented dataset: {len(augmented_abstracts)} samples")

# Show example of synthetic abstract
print(f"\\nExample synthetic abstract:")
print(f"'{augmented_abstracts[-1]}'")

# Prepare augmented training dataset
augmented_dataset = prepare_dataset_for_training(augmented_abstracts, tokenizer, max_length=256)

# Modified training configuration
modified_config = {
    'learning_rate': 1e-5,        # Lower learning rate
    'num_epochs': 5,              # More epochs  
    'batch_size': 2,              # Same batch size
    'warmup_steps': 20,           # More warmup
    'max_grad_norm': 0.5,         # Tighter gradient clipping
    'save_steps': 50,
    'logging_steps': 5,
}

print(f"\\nModified Training Configuration:")
for key, value in modified_config.items():
    print(f"  {key}: {value}")

# Create fresh model copy for modified training
modified_model = copy.deepcopy(model)
modified_model.train()

# Setup modified training arguments
modified_args = TrainingArguments(
    output_dir='./modified_model',
    overwrite_output_dir=True,
    num_train_epochs=modified_config['num_epochs'],
    per_device_train_batch_size=modified_config['batch_size'],
    learning_rate=modified_config['learning_rate'],
    warmup_steps=modified_config['warmup_steps'],
    logging_steps=modified_config['logging_steps'],
    save_steps=modified_config['save_steps'],
    max_grad_norm=modified_config['max_grad_norm'],
    no_cuda=True,
    dataloader_pin_memory=False,
    remove_unused_columns=False,
    report_to=None,
    logging_dir=None,
    save_total_limit=1,
)

# Create modified trainer
modified_trainer = Trainer(
    model=modified_model,
    args=modified_args,
    train_dataset=augmented_dataset,
    data_collator=data_collator,
    tokenizer=tokenizer,
)

print(f"\\n=== STARTING MODIFIED TRAINING ===")
print(f"Training on {len(augmented_dataset)} samples for {modified_config['num_epochs']} epochs")

# Train the modified model
modified_start_time = time.time()

try:
    modified_result = modified_trainer.train()
    modified_end_time = time.time()
    modified_training_time = modified_end_time - modified_start_time
    
    print(f"\\n=== MODIFIED TRAINING COMPLETED ===")
    print(f"Training time: {modified_training_time:.2f} seconds ({modified_training_time/60:.2f} minutes)")
    print(f"Final training loss: {modified_result.training_loss:.4f}")
    
    modified_successful = True
    
except Exception as e:
    print(f"Modified training failed: {e}")
    modified_end_time = time.time()
    modified_training_time = modified_end_time - modified_start_time
    modified_successful = False

print(f"\\nModified training summary:")
print(f"- Duration: {modified_training_time:.2f} seconds") 
print(f"- Success: {modified_successful}")
if modified_successful:
    print(f"- Final loss: {modified_result.training_loss:.4f}")
    print(f"- Steps completed: {modified_result.global_step}")
```
